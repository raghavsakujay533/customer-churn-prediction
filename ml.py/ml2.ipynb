```python
import pandas as pd
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as pl

```

```python
df=pd.read_csv("linedata1.csv")
df
```

```python
x=df[["salary"]]
y=df.expenses
```

```python
pl.plot(x,y,color="red",marker="+")
pl.xlabel("--x--axis")
pl.ylabel("--y--axis")
pl.title("yeh ek plot hai")
```

```python
model=LinearRegression()
model.fit(x,y)
```

```python
model.predict([[40000]])
```

```python
m=model.coef_
print(m)
```

```python
b=model.intercept_
print(b)
```

```python
y=(m*40000)+b   
```

```python
print(y)
```

```python
pl.ylabel(b)
pl.xlabel(m)
b
m
pl.show
```
