```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn . linear_model import LinearRegression
```

```python
df=pd.read_csv("homeprices.csv")
df
```

```python
x=df[["area"]]
y=df["price"]
```

```python
plt.scatter(x,y, color="grey",marker="o")
plt.title("table showing ")
plt.xlabel("area""--x--axis")
plt.ylabel("price""--y--axis")
```

```python
model=LinearRegression()
model.fit(x,y)
```

```python
model.predict([[3100]])
```

```python
m=model.coef_
print(m)
```

```python
b=model.intercept_
print(b)
```

```python
y=(m*3100)+b
```

```python
print(y)
```
